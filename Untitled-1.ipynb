```python
def resumetable(df):
    print(f"Dataset Shape: {df.shape}")
    summary = pd.DataFrame(df.dtypes, columns=['dtypes'])
    summary = summary.reset_index()
    summary['Name'] = summary['index']
    summary = summary[['Name', 'dtypes']]
    summary['Missing'] = df.isnull().sum().values    
    summary['Uniques'] = df.nunique().values
    summary['First Value'] = df.loc[0].values
    summary['Second Value'] = df.loc[1].values
    summary['Third Value'] = df.loc[2].values
    return summary
```

```python
from scipy import stats
import numpy as np
import pandas as pd

def ttest_onesided(data, alpha=0.01):
    data_clean = pd.to_numeric(data, errors='coerce').dropna()
    
    pop_mean = data_clean.mean()
    sample_data = data_clean.sample(min(500, len(data_clean)), random_state=42)
    sample_mean = sample_data.mean()
    
    t_stat, p_val = stats.ttest_1samp(sample_data, pop_mean)
    
    print(f"Mean of Population: {pop_mean}")
    print(f"Mean of Sample: {sample_mean}")
    print(f"t-test value: {t_stat}")
    print("Comparing p_value by...\n")
    print(f"p-value result: {p_val}")
    
    if p_val < alpha:
        print("we are rejecting null hypothesis")
    else:
        print("we fail to reject null hypothesis")
```

```python
goal_col = 'usd_goal_real' if 'usd_goal_real' in df_kick.columns else 'goal'

ttest_onesided(np.log(df_kick[goal_col] + 1))
```

```python
from scipy import stats
import numpy as np

def ttest_twosided(samples, alpha=0.01):
    sample1, sample2 = samples[0], samples[1]
    
    mean1 = np.mean(sample1)
    mean2 = np.mean(sample2)
    
    t_stat, p_val = stats.ttest_ind(sample1, sample2, alternative='two-sided')
    
    print(f"Mean of Sample 1: {mean1}")
    print(f"Mean of Sample 2: {mean2}")
    print(f"t-test value: {t_stat}")
    print(f"p-value result: {p_val}")
    
    if p_val < alpha:
        print("we are rejecting null hypothesis")
    else:
        print("we fail to reject null hypothesis")
```

```python
import pandas as pd
from scipy.stats import chi2_contingency, chi2

def chi2_test(cols, alpha=0.05):
    contingency_table = pd.crosstab(df_kick[cols[0]], df_kick[cols[1]])
    
    
    stat, p, dof, expected = chi2_contingency(contingency_table)
    
    prob = 1 - alpha
    critical = chi2.ppf(prob, dof)
    
    print("CHI-SQUARED TEST:")
    print(f"dof={dof}, probability={prob}, critical={critical:.4f}, stat={stat:.4f}")
    print("Accept or Reject H0:")
    
    if abs(stat) >= critical:
        print("Dependent (reject H0)")
    else:
        print("Independent (fail to reject H0)")
```
